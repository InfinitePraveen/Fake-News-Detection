# 07 - Final Model Export
## Fake News Detection - Export for Web App

```python
import pandas as pd
import numpy as np
import joblib
import json
import warnings
warnings.filterwarnings('ignore')
```

```python
# Load best model
best_model = joblib.load('../webapp/model/best_logistic_regression.pkl')
print("✅ Model loaded successfully!")
```

```python
# Load TF-IDF vectorizer
tfidf = joblib.load('../webapp/model/tfidf_vectorizer.pkl')
print("✅ TF-IDF vectorizer loaded successfully!")
```

```python
# Create metadata
metadata = {
    'model_name': 'Logistic Regression',
    'model_version': '2.0',
    'features': tfidf.get_feature_names_out().tolist(),
    'n_features': len(tfidf.get_feature_names_out()),
    'coef': best_model.coef_[0].tolist(),
    'intercept': best_model.intercept_[0],
    'classes': ['Fake', 'Real'],
    'parameters': best_model.get_params(),
    'dataset': 'WELFake'
}

# Save metadata
with open('../webapp/model/model_metadata.json', 'w') as f:
    json.dump(
        metadata,
        f,
        indent=4,
        default=lambda value: value.item() if isinstance(value, np.generic) else str(value)
    )
print("✅ Metadata saved to '../webapp/model/model_metadata.json'")
```

```python
# Export model for web app
webapp_model_path = '../webapp/model/fake_news_model.pkl'
joblib.dump(best_model, webapp_model_path)
print(f"✅ Model exported to '{webapp_model_path}'")
```

```python
# Export TF-IDF vectorizer for web app
webapp_vectorizer_path = '../webapp/model/tfidf_vectorizer.pkl'
joblib.dump(tfidf, webapp_vectorizer_path)
print(f"✅ Vectorizer exported to '{webapp_vectorizer_path}'")
```

```python
# Test exported model
test_text = "This is a test news article about climate change."
test_tfidf = tfidf.transform([test_text])
prediction = best_model.predict(test_tfidf)[0]
probability = best_model.predict_proba(test_tfidf)[0]

print("="*50)
print("TEST EXPORTED MODEL")
print("="*50)
print(f"Text: {test_text}")
print(f"Prediction: {'Real' if prediction == 1 else 'Fake'}")
print(f"Confidence: {max(probability):.2%}")
print(f"Fake Probability: {probability[0]:.2%}")
print(f"Real Probability: {probability[1]:.2%}")
```

```python
# Create final summary
summary = {
    'project_name': 'Fake News Detection',
    'model_type': 'Logistic Regression',
    'vectorizer': 'TF-IDF',
    'total_features': len(tfidf.get_feature_names_out()),
    'model_parameters': best_model.get_params(),
    'classes': ['Fake', 'Real'],
    'export_date': pd.Timestamp.now().strftime('%Y-%m-%d %H:%M:%S')
}

with open('../webapp/model/export_summary.json', 'w') as f:
    json.dump(summary, f, indent=4)
print("\n✅ Export summary saved to '../webapp/model/export_summary.json'")
print("\n📦 ALL FILES EXPORTED SUCCESSFULLY FOR WEB APP DEPLOYMENT!")
```
