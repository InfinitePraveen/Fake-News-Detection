# 04 - Model Training
## Fake News Detection - Logistic Regression

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import joblib
import warnings
warnings.filterwarnings('ignore')
```

```python
# Load preprocessed data
train_data = np.load('../data/train_tfidf.npz')
test_data = np.load('../data/test_tfidf.npz')

X_train = train_data['arr_0']
X_test = test_data['arr_0']

# Load labels
import pickle
with open('../data/train_test_data.pkl', 'rb') as f:
    data = pickle.load(f)
    y_train = data['y_train']
    y_test = data['y_test']

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
# Train Logistic Regression model
model = LogisticRegression(
    C=1.0,
    max_iter=1000,
    random_state=42,
    class_weight='balanced'
)

model.fit(X_train, y_train)
print("✅ Model training completed!")
```

```python
# Predictions
y_pred = model.predict(X_test)
y_pred_proba = model.predict_proba(X_test)

# Accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"\nTest Accuracy: {accuracy:.4f}")
```

```python
# Classification Report
print("\n" + "="*60)
print("CLASSIFICATION REPORT")
print("="*60)
print(classification_report(y_test, y_pred, target_names=['Fake', 'Real']))
```

```python
# Feature importance
feature_importance = pd.DataFrame({
    'feature': range(X_train.shape[1]),
    'importance': np.abs(model.coef_[0])
}).sort_values('importance', ascending=False).head(20)

print("\nTop 20 Most Important Features:")
print(feature_importance)
```

```python
# Save model
joblib.dump(model, '../webapp/model/logistic_regression_model.pkl')
print("\n✅ Model saved to '../models/logistic_regression_model.pkl'")
```
